```python
from pathlib import Path
import os

print("Home:")
print(Path.home())

print("\nMounted directories:")
for p in Path("/").iterdir():
    print(p)
```

```python
import os

print("\nEnvironment variables containing VS Code:")
for k, v in os.environ.items():
    if "VSCODE" in k.upper() or "COLAB" in k.upper():
        print(k, "=", v)
```
